```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

from google.colab import files
uploaded= files.upload()
```

```python
data=pd.read_csv('smart_grid_stability_augmented.csv')
```

```python
x=data.drop(['stab','stabf'],axis=1)
y = data['stabf'].map({'stable': 1, 'unstable': 0})

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)

scaler= StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled= scaler.fit_transform(x_test)

dt_model = DecisionTreeClassifier()
dt_model.fit(x_train_scaled, y_train)

# Step 3: Predict and calculate accuracy
y_pred = dt_model.predict(x_test_scaled)


# Accuracy
accuracy = accuracy_score(y_test, y_pred)

# Precision, Recall, F1-Score
report = classification_report(y_test, y_pred)

print(f"Accuracy of the DT model: {accuracy * 100:.2f}%")
print(f"Classification Report:\n{report}")
```
